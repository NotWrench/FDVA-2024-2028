```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv")
df.head()
```

```python
plt.plot(df["Sales"], marker='o', label="Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Line Chart")
plt.legend()
plt.show()
```

```python
plt.bar(df["Category"], df["Sales"])
plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Bar Chart")
plt.show()
```

```python
plt.hist(df["Value"], bins=20, edgecolor='black')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
